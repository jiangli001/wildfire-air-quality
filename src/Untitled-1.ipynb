```python
import pandas as pd
```

```python
df = pd.read_csv("../data/fire_incidents.csv")
df = df[[
    "incident_latitude",
    "incident_longitude",
    "incident_dateonly_created",
    "incident_dateonly_extinguished",
    "SiteName"
]]
df.columns = ["latitude", "longitude", "start_date", "end_date", "site_name"]
# filter out rows with missing values
df = df.dropna(subset=["latitude", "longitude", "start_date", "end_date"])
df.drop_duplicates(inplace=True)
```

```python
mapping = pd.read_csv("../data/site_name_mapping.csv")
```

```python
df_merged = df.merge(mapping, left_on="site_name", right_on='name',how="left")
```

```python
df_merged = df_merged[df_merged['site'].notnull()]
```

```python
df_merged.to_csv("../data/processed_fire_incidents.csv", index=False)
```

```python
weather = pd.read_csv("../data/weather_data_new.csv")
```

```python
weather['day'] = pd.to_datetime(weather['time']).dt.date
```

```python
weather['day_hour'] = pd.to_datetime(weather['time']).dt.hour
```

```python
pm25 = pd.read_csv("../data/processed_pm25.csv")
```

```python
pm25 = pm25.rename(columns={"site": "site_id", 'date': 'day', 'start_hour': 'day_hour'})
```

```python
pm25['day'] = pd.to_datetime(pm25['day']).dt.date
```

```python
merged_all = pm25.merge(weather, on=['site_id', 'day', 'day_hour'], how='inner')
```

```python
merged_all.to_csv("../data/final_merged_data_new.csv", index=False)
```
